```python
daily_paid_events = (
    df[df["payment_status"] == "Paid"]
    .groupby(["customer_id", "payment_date"])
    .agg(
        total_delay=("payment_delay", "sum"),
        paid_count=("payment_delay", "count")
    )
    .reset_index()
    .sort_values(["payment_date", "customer_id"])
)

daily_paid_events["cumulative_delay"] = (
    daily_paid_events.groupby("customer_id")["total_delay"].cumsum()
)

daily_paid_events["cumulative_paid_count"] = (
    daily_paid_events.groupby("customer_id")["paid_count"].cumsum()
)
```

```python
left = df[["customer_id", "invoice_date"]].copy()
left["row_id"] = np.arange(len(df))

left = left.sort_values("invoice_date")

right = daily_paid_events[
    ["customer_id", "payment_date", "cumulative_delay", "cumulative_paid_count"]
].sort_values("payment_date")

history = pd.merge_asof(
    left,
    right,
    left_on="invoice_date",
    right_on="payment_date",
    by="customer_id",
    direction="backward",
    allow_exact_matches=False
)
```

```python
history["previous_avg_delay"] = (
    history["cumulative_delay"] /
    history["cumulative_paid_count"]
)

history = history.sort_values("row_id")

df["previous_avg_delay"] = history["previous_avg_delay"].to_numpy()
```
